# ✨ Getting Started

This tutorial covers the basics of getting started with `pyhctsa` including:

- Getting up and running with feature extraction on a single time series
- Extracting features from datasets of many instances
- Specifying custom labels for results tables
- Calling individual feature functions
- Parallel computing

# 📘 (1) Simple Feature Extraction

Simple feature extraction

```python
data = get_dataset() # load the empirical1000 dataset from the pyhctsa utils
print(type(data), type(data[0]))
print(data[0])
```

Instantiate a calculator with the full hctsa feature set

```python
calc = FeatureCalculator()
```

Extract features from a single time-series instance

```python
res = calc.extract(data[0])
```

`pyhctsa` will return a pandas DataFrame corresponding to the feature matrix:

```python
res
```

Here, each row corresponds to a time-series instance (in this case, a single row since features were extracted for a single time-series instance), and each column corresponds to a time-series feature.

### 📕 Multiple Time-Series Instances

Lets now extract features for the first 10 time-series instances in the `empirical1000` dataset. Since the input can be inhomogeneous (i.e., each time-series instance may differ in length), we must pass a list of numpy arrays (each array corresponding to a time-series instance):

```python
data10 = data[:10] # take the first 10 time-series instances from empirical1000
print(type(data10), type(data10[0]))
```

```python
res10 = calc.extract(data10)
```

```python
res10
```

By default, `pyhctsa` will assign a generic label to index each time-series instance (row) in the results table. For example, `ts_1` corresponds to the first time-series instance passed into the calculator, `ts_2` to the second, etc. To inspect the results for a specific instance (a feature vector), we can use the pandas method `.loc`:

```python
# inspect the results for ts_1
res10.loc['ts_1']
```

### 🏷 Custom time-series labels

If you would like to assign custom labels to each time-series instance, you can pass a list of strings along with the data when extracting features:

```python
labels = [f'e1000_{x}' for x in range(1, 11)]
res10_labelled = calc.extract(data10[:10], labels=labels)
```

```python
res10_labelled
```

Here, we can see that each time-series instance is now indexed by its custom label e.g., `e1000_1`, `e1000_2`, which we can now use to inspect results:

```python
res10_labelled.loc['e1000_9']
```

# 🤖 (2) Additional Functionality

## ⚙️ Calling individual feature functions

While the `FeatureCalculator` massively simplifies the extraction of thousands of featues with a single line of code, `pyhctsa` also allows users to call individual feature functions from their respective modules, located in `pyhctsa.operations.<modlue_name>`. 

```python
import pkgutil
import pyhctsa.operations as ops
import inspect
mods = [m.name for m in pkgutil.iter_modules(ops.__path__, ops.__name__ + ".")]
print("\n".join(mods))
```

For example, we can call the rescaled auto-density (RAD) function by importing it from the `pychtsa.operations.criticality` module:

```python
from pyhctsa.operations.criticality import rad # import rad from the criticality module...
```

We can then call `rad` directly on our data, choosing whatever parameters we need. For details on what the function computes and how to configure it, refer to its docstring.

*NOTE:* Generic feature functions expect a 1D array-like input (a single time series). Also be careful when calling functions manually: make sure the data you pass in __matches the preprocessing assumptions for your use case__. In particular, decide explicitly whether you want to supply raw data or z-scored/standardized data as this choice can substantially influence the resulting feature values.

```python
print(inspect.getdoc(rad))
```

```python
rad_result = rad(data[0], tau=1, centre=True) # pass in the raw data (not z-scored)
print(rad_result)
```

For feature functions which expect/assume a z-scored input, you can first pre-process the data using the `z_score` function:

```python
from pyhctsa.utils import z_score
```

```python
z_score(data[0]) # z-score normalise the data along the time axis
```

## 🏗️ Parallel Computing

Computing large feature sets across many time series can be computationally expensive. Since feature extraction is independent across individual time series, the task is naturally suited to parallelisation. `pyhctsa` supports parallel execution using the pathos library, distributing batches of time series across multiple worker processes to reduce overall runtime.

#### Setting up a local parallel distributor

```python
# import the LocalDistributor
from pyhctsa.distribute import LocalDistributor
dist = LocalDistributor(n_workers=8) # use up to the number of physical CPU cores available on the machine
```

To distribute feature computation, pass the distributor instance to the extract method:

```python
calc = FeatureCalculator()
res_distributed = calc.extract(data10, distributor=dist)
```

Parallelisation introduces some overhead from process creation and inter-process communication, so speedups may be modest for very small datasets. However, for larger collections of time series, distributing the workload can substantially reduce total computation time. Even in this small example (10 time series), parallel execution reduced runtime by roughly a factor of three compared to sequential feature extraction.
